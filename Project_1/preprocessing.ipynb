# 3.2 Chuẩn hóa dữ liệu

```python
df.info()
```

```python
import numpy as np
import pandas as pd

# Hàm chuyển đổi "Giá" (về đơn vị triệu đồng)
def clean_gia(value):
    if pd.isna(value):
        return np.nan
    value = str(value).replace('.', '').replace('đ', '').replace(' ', '')
    try:
        return float(value) / 1_000_000  # Chia cho 10^6 để ra triệu
    except:
        return np.nan

# Hàm chuyển đổi "Khoảng giá min/max" (đã là triệu, chỉ cần chuẩn hóa)
def clean_khoang_gia(value):
    if pd.isna(value):
        return np.nan
    value = str(value).lower().replace('tr', '').replace(',', '.').strip()
    try:
        return float(value)  # Vì "tr" đã là triệu
    except:
        return np.nan

# Áp dụng cho các cột
df['Giá'] = df['Giá'].apply(clean_gia)
df['Khoảng giá min'] = df['Khoảng giá min'].apply(clean_khoang_gia)
df['Khoảng giá max'] = df['Khoảng giá max'].apply(clean_khoang_gia)
```

## 3.2 Handle Null Values

### 3.2.1 Thống kê null values

```python
# Thống kê null
missing = df.isnull().sum().reset_index()
missing.columns = ['Cột', 'Số lượng Null']
missing['Tỉ lệ (%)'] = (missing['Số lượng Null'] / len(df)) * 100

# Vẽ biểu đồ
plt.figure(figsize=(10,5))
bars = plt.bar(missing['Cột'], missing['Tỉ lệ (%)'], color='skyblue')

# Thêm nhãn số lượng null trên mỗi thanh bar
for bar, count in zip(bars, missing['Số lượng Null']):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.2,   # vị trí
             f'{count}', ha='center', va='bottom', fontsize=10, color='black')

plt.title('Tỉ lệ và số lượng giá trị Null trong từng cột')
plt.ylabel('Tỉ lệ Null (%)')
plt.xlabel('Tên cột')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# ==> Xóa luôn các hàng chứa null của ['Giá', 'Thương hiệu', 'Dòng xe']
```

```python
# Kiểm tra giá trị null
null_diff = df[df['Chênh lệch giá'].isna()]
print(f"Số lượng hàng bị NaN trong 'Chênh lệch giá': {len(null_diff)}")
```

```python
import numpy as np

# 1️ Tính các giá trị trung vị cần thiết
median_diff = df['Chênh lệch giá'].median()
median_min = df['Khoảng giá min'].median()
median_max = df['Khoảng giá max'].median()

print("Trung vị Khoảng giá min:", median_min)
print("Trung vị Khoảng giá max:", median_max)
print("Trung vị Chênh lệch giá:", median_diff)

# 2️ Trường hợp 1: Thiếu max -> ước lượng bằng min + median_diff
df.loc[df['Khoảng giá max'].isna() & df['Khoảng giá min'].notna(), 'Khoảng giá max'] = \
    df['Khoảng giá min'] + median_diff

# 3️ Trường hợp 2: Thiếu min -> ước lượng bằng max - median_diff
df.loc[df['Khoảng giá min'].isna() & df['Khoảng giá max'].notna(), 'Khoảng giá min'] = \
    df['Khoảng giá max'] - median_diff

# 4️ Trường hợp 3: Thiếu cả hai -> gán trung vị cho cả hai
df.loc[df['Khoảng giá min'].isna() & df['Khoảng giá max'].isna(), ['Khoảng giá min', 'Khoảng giá max']] = \
    [median_min, median_max]

# 5️ Cập nhật lại chênh lệch giá
df['Chênh lệch giá'] = df['Khoảng giá max'] - df['Khoảng giá min']

# 6️ Kiểm tra lại
print("\nSố lượng null sau khi xử lý:")
print(df[['Khoảng giá min', 'Khoảng giá max', 'Chênh lệch giá']].isna().sum())
```
